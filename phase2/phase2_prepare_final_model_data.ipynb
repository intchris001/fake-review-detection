# 1. Loading Phase 1 final_dataset

```python
import pandas as pd

df = pd.read_csv("../phase1/final_dataset.csv")

print("Original shape:", df.shape)
```

# 2. Deleting dangerous variables (columns)

```python
leakage_cols = [
    "fake_confidence_score",
    "sent_is_fake_review"
]

df_clean = df.drop(columns=leakage_cols, errors="ignore")

print("After removing leakage columns:", df_clean.shape)
```

# 3. Checking others

```python
print(df_clean.columns.tolist())
```

# 4. Save clean dataset

```python
df_clean.to_csv("clean_model_dataset.csv", index=False)

print("Clean modeling dataset saved.")
```
